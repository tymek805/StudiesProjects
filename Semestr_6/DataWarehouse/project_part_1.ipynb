```python
import pandas as pd
from ydata_profiling import ProfileReport

df = pd.read_csv('globalterrorismdb.csv', encoding='ISO-8859-1', low_memory=False)
df = df[df['iyear'] >= 1997]
# Stay:
# eventid, iyear, imonth, iday, _, country_txt
# Remove:
# approxdate, extended, resolution, country, _,

df = df[
    ["eventid", "iyear", "imonth", "iday", "country_txt", "region_txt", "city", "attacktype1_txt", "targtype1_txt",
     "gname", "nkill", "nwound", "success", "suicide"]]
df.to_csv('processed_data.csv', index=False)
profile = ProfileReport(df, title="Profiling Report")

profile.to_file("processed_data.html")
```
